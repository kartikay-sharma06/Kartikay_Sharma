```python
#to drop rows
df.dropna(subset=['embarked'],inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['sex']=le.fit_transform(df['sex'])
df['embarked']=le.fit_transform(df['embarked'])
```

```python
df.head()
```

```python
df.astype(int)
```

```python
x=df.drop('survived',axis=1)
y=df['survived']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
(accuracy_score(y_test,y_pred))
```

```python
(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

KNN

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train_scale=sc.fit_transform(x_train)
x_test_scale=sc.transform(x_test)
```

```python
(x_train_scale)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scale,y_train)
```

```python
y_prediction_knn=knn.predict(x_test_scale)
```

```python
accuracy_score(y_test,y_prediction_knn)
```

```python
confusion_matrix(y_test,y_prediction_knn)
```

```python
print(classification_report(y_test,y_prediction_knn))
```

Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(x_train,y_train)
```

```python
nb_pred=nb.predict(x_test)
```

```python
accuracy_score(y_test,nb_pred)
```

```python
confusion_matrix(y_test,nb_pred)
```

```python
print(classification_report(y_test,nb_pred))
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(random_state=42)
dt.fit(x_train_scale,y_train)
```

```python
y_predict_dt=dt.predict(x_test_scale)
```

```python
accuracy_score(y_test,y_predict_dt)
```

```python
confusion_matrix(y_test,y_predict_dt)
```

```python
print(classification_report(y_test,y_predict_dt))
```
